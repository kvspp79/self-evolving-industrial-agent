```python
# ================================================================
# A SELF-EVOLVING EXPLAINABLE RATIONAL AGENT FOR ADAPTIVE
# INDUSTRIAL DECISION INTELLIGENCE USING STATISTICAL TREND LEARNING
# AI4I 2020 PREDICTIVE MAINTENANCE DATASET
# ================================================================

# If required, install once in a separate terminal/notebook cell:
# !pip install pandas numpy matplotlib seaborn scikit-learn xgboost shap

import os
import re
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report
)

warnings.filterwarnings("ignore")
np.random.seed(42)

# ================================================================
# 1. LOAD DATASET
# ================================================================

DATA_PATH = r"C:\Users\kvspp\Downloads\aiml project\ai4i+2020+predictive+maintenance+dataset\ai4i2020.csv"

if not os.path.exists(DATA_PATH):
    DATA_PATH = "ai4i2020.csv"

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        "AI4I dataset not found. Change DATA_PATH at the top of this cell "
        "to the location of ai4i2020.csv."
    )

df = pd.read_csv(DATA_PATH)

print("=" * 70)
print("AI4I 2020 PREDICTIVE MAINTENANCE DATASET")
print("=" * 70)
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nFailure distribution:")
print(df["Machine failure"].value_counts())
print("\nFailure percentage:")
print((df["Machine failure"].value_counts(normalize=True) * 100).round(3))

# ================================================================
# 2. BASIC EDA
# ================================================================

print("\nFIRST 5 RECORDS")
display(df.head())

print("\nDATA TYPES")
print(df.dtypes)

print("\nMISSING VALUES")
print(df.isnull().sum())

# Machine type distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Type")
plt.title("Machine Type Distribution")
plt.xlabel("Machine Type")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# Failure distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Machine failure")
plt.title("Machine Failure Distribution")
plt.xlabel("Machine Failure")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# Numeric distributions
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
eda_cols = [
    "Air temperature [K]", "Process temperature [K]",
    "Rotational speed [rpm]", "Torque [Nm]", "Tool wear [min]"
]
for ax, col in zip(axes.ravel(), eda_cols):
    ax.hist(df[col], bins=30)
    ax.set_title(col)
    ax.set_xlabel(col)
    ax.set_ylabel("Frequency")
axes.ravel()[-1].axis("off")
plt.tight_layout()
plt.show()

# Correlation heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

# ================================================================
# 3. DATA PREPARATION / LEAKAGE CONTROL
# ================================================================

# Failure-mode columns are excluded because they directly describe failure causes.
# UDI and Product ID are identifiers, not predictive machine measurements.
DROP_COLUMNS = [
    "Machine failure", "UDI", "Product ID",
    "TWF", "HDF", "PWF", "OSF", "RNF"
]

X = df.drop(columns=DROP_COLUMNS, errors="ignore").copy()
y = df["Machine failure"].astype(int).copy()

# One-hot encode Type; drop first category to avoid redundant dummy variables.
X = pd.get_dummies(X, columns=["Type"], drop_first=True, dtype=int)

# Keep the original row indices so trend records can be aligned later.
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# XGBoost requires simple feature names.
def clean_name(name):
    return re.sub(r"[^A-Za-z0-9_]", "_", str(name))

X_train = X_train.copy()
X_test = X_test.copy()
X_train.columns = [clean_name(c) for c in X_train.columns]
X_test.columns = X_train.columns

print("\nFINAL FEATURES")
print(X_train.columns.tolist())
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

negative = int((y_train == 0).sum())
positive = int((y_train == 1).sum())
scale_pos_weight = negative / max(positive, 1)
print("Training normal:", negative)
print("Training failures:", positive)
print("Scale pos weight:", round(scale_pos_weight, 4))

# ================================================================
# 4. MAIN XGBOOST MODEL
# ================================================================

model = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

failure_probability = model.predict_proba(X_test)[:, 1]
y_pred = (failure_probability >= 0.50).astype(int)

print("\n===== MAIN MODEL PERFORMANCE =====")
print("Accuracy :", round(accuracy_score(y_test, y_pred) * 100, 2), "%")
print("Precision:", round(precision_score(y_test, y_pred, zero_division=0) * 100, 2), "%")
print("Recall   :", round(recall_score(y_test, y_pred, zero_division=0) * 100, 2), "%")
print("F1-score :", round(f1_score(y_test, y_pred, zero_division=0) * 100, 2), "%")
print("ROC-AUC  :", round(roc_auc_score(y_test, failure_probability) * 100, 2), "%")
print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.title("XGBoost Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()

# ================================================================
# 5. EXPLAINABLE AI WITH SHAP
# ================================================================

try:
    import shap

    explainer = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X_test)

    print("\n===== SHAP FEATURE IMPORTANCE =====")
    shap.summary_plot(shap_values, X_test, show=True)
    shap.summary_plot(shap_values, X_test, plot_type="bar", show=True)

    # Explain the first test record.
    try:
        shap.waterfall_plot(
            shap.Explanation(
                values=shap_values[0],
                base_values=explainer.expected_value,
                data=X_test.iloc[0].values,
                feature_names=X_test.columns.tolist()
            ),
            max_display=10,
            show=True
        )
    except Exception:
        pass

    print("Example test-record failure probability:", round(failure_probability[0] * 100, 2), "%")
except ImportError:
    print("SHAP is not installed. Run: pip install shap")

# ================================================================
# 6. RATIONAL DECISION AGENT
# ================================================================

# Original project policy: probability -> operational action.
POLICIES = {
    "Original (30%, 60%, 80%)": (0.30, 0.60, 0.80),
    "Best F1 (35%, 65%, 85%)": (0.35, 0.65, 0.85),
    "Best Cost (25%, 55%, 80%)": (0.25, 0.55, 0.80),
    "Lower Thresholds (20%, 50%, 75%)": (0.20, 0.50, 0.75)
}

ACTION_ORDER = ["Normal", "Monitor", "Schedule Maintenance", "Critical"]


def make_decisions(probabilities, thresholds):
    t1, t2, t3 = thresholds
    actions = np.empty(len(probabilities), dtype=object)
    actions[probabilities < t1] = "Normal"
    actions[(probabilities >= t1) & (probabilities < t2)] = "Monitor"
    actions[(probabilities >= t2) & (probabilities < t3)] = "Schedule Maintenance"
    actions[probabilities >= t3] = "Critical"
    return actions


def pretty_decision(action):
    return {
        "Normal": "NORMAL - Continue Operation",
        "Monitor": "MONITOR - Increase Monitoring",
        "Schedule Maintenance": "SCHEDULE MAINTENANCE",
        "Critical": "CRITICAL - Immediate Inspection"
    }[action]

original_actions = make_decisions(
    failure_probability, POLICIES["Original (30%, 60%, 80%)"]
)

agent_results = pd.DataFrame({
    "Actual Failure": y_test.to_numpy(),
    "Failure Probability": failure_probability,
    "Agent Decision": [pretty_decision(a) for a in original_actions]
}, index=X_test.index)

agent_summary = agent_results.groupby("Agent Decision").agg(
    Total_Machines=("Actual Failure", "count"),
    Actual_Failures=("Actual Failure", "sum"),
    Average_Failure_Probability=("Failure Probability", "mean"),
    Actual_Failure_Rate=("Actual Failure", "mean")
).reset_index()

agent_summary["Average_Failure_Probability"] *= 100
agent_summary["Actual_Failure_Rate"] *= 100

print("\n===== RATIONAL AGENT EVALUATION =====")
display(agent_summary.round(2))

# ================================================================
# 7. STATISTICAL TREND LEARNING
# ================================================================

trend_df = df.copy()
trend_df = trend_df.sort_values("UDI").reset_index(drop=True)

sensor_columns = {
    "Torque": "Torque [Nm]",
    "Tool Wear": "Tool wear [min]",
    "Air Temperature": "Air temperature [K]",
    "Process Temperature": "Process temperature [K]",
    "Rotational Speed": "Rotational speed [rpm]"
}

window = 20

for name, col in sensor_columns.items():
    values = pd.to_numeric(trend_df[col], errors="coerce")
    trend_df[f"{name}_Moving_Average"] = values.rolling(window, min_periods=1).mean()
    trend_df[f"{name}_EWMA"] = values.ewm(span=window, adjust=False).mean()
    trend_df[f"{name}_Trend_Slope"] = values.rolling(window, min_periods=window).apply(
        lambda x: np.polyfit(np.arange(len(x)), x, 1)[0], raw=True
    )

trend_df["Torque_Trend_State"] = np.select(
    [
        trend_df["Torque_Trend_Slope"] > 0.05,
        trend_df["Torque_Trend_Slope"] < -0.05
    ],
    ["Increasing", "Decreasing"],
    default="Stable"
)

print("\n===== TORQUE TREND STATES =====")
print(trend_df["Torque_Trend_State"].value_counts(dropna=False))

plt.figure(figsize=(12, 5))
plt.plot(trend_df["UDI"], trend_df["Torque [Nm]"], alpha=0.25, label="Torque")
plt.plot(trend_df["UDI"], trend_df["Torque_Moving_Average"], label="Moving Average")
plt.plot(trend_df["UDI"], trend_df["Torque_EWMA"], label="EWMA")
plt.title("Statistical Trend Learning - Torque")
plt.xlabel("UDI")
plt.ylabel("Torque (Nm)")
plt.legend()
plt.tight_layout()
plt.show()

# Align trend records with the original test rows using UDI.
test_udi = df.loc[X_test.index, "UDI"].to_numpy()
trend_lookup = trend_df.set_index("UDI")
test_trends = trend_lookup.loc[test_udi].reset_index()

assert len(test_trends) == len(X_test)
assert np.array_equal(test_trends["UDI"].to_numpy(), test_udi)

agent_results = agent_results.reset_index(drop=True)
agent_results["Torque_Trend_State"] = test_trends["Torque_Trend_State"].to_numpy()

# ================================================================
# 8. TREND-AWARE RATIONAL AGENT
# ================================================================

def trend_aware_action(original_action, trend_state):
    level = ACTION_ORDER.index(original_action)
    if trend_state == "Increasing":
        level = min(level + 1, len(ACTION_ORDER) - 1)
    return ACTION_ORDER[level]

trend_actions = [
    trend_aware_action(a, t)
    for a, t in zip(original_actions, test_trends["Torque_Trend_State"])
]

trend_agent_results = agent_results.copy()
trend_agent_results["Trend_Aware_Decision"] = trend_actions

print("\n===== TREND-AWARE RATIONAL AGENT =====")
print(pd.Series(trend_actions).value_counts())

# Diagnostic: show how aggressive the increasing-trend escalation is.
trend_diagnostic = pd.DataFrame({
    "Actual Failure": y_test.to_numpy(),
    "Trend": test_trends["Torque_Trend_State"].to_numpy(),
    "Escalated": np.array(trend_actions) != np.array(original_actions)
})
print("\nTrend diagnostic:")
display(
    trend_diagnostic.groupby("Trend").agg(
        Records=("Actual Failure", "size"),
        Failures=("Actual Failure", "sum"),
        Escalated=("Escalated", "sum")
    ).reset_index()
)

# ================================================================
# 9. INDEPENDENT VALIDATION SET FOR POLICY ANALYSIS
# ================================================================

# The policy-selection/feedback experiments are performed on a validation
# split of the training data, keeping the original test set for model reporting.
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train, y_train,
    test_size=0.25,
    random_state=42,
    stratify=y_train
)

val_positive = int((y_fit == 1).sum())
val_negative = int((y_fit == 0).sum())

validation_model = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=val_negative / max(val_positive, 1),
    random_state=42,
    n_jobs=-1
)
validation_model.fit(X_fit, y_fit)
val_prob = validation_model.predict_proba(X_val)[:, 1]

# Align UDI and trends for validation rows.
val_udi = df.loc[X_val.index, "UDI"].to_numpy()
val_trends = trend_lookup.loc[val_udi].reset_index()

validation_results = pd.DataFrame({
    "Actual Failure": y_val.to_numpy(),
    "Failure Probability": val_prob,
    "Torque_Trend_State": val_trends["Torque_Trend_State"].to_numpy()
}, index=X_val.index).reset_index(drop=True)

validation_results["Agent Decision"] = [
    pretty_decision(a) for a in make_decisions(val_prob, POLICIES["Original (30%, 60%, 80%)"])
]

print("\nValidation set:", len(validation_results), "records")
print("Validation failures:", int(validation_results["Actual Failure"].sum()))

# ================================================================
# 10. TREND ESCALATION THRESHOLD COMPARISON
# ================================================================


def evaluate_alerts(actual, predicted):
    actual = np.asarray(actual).astype(int)
    predicted = np.asarray(predicted).astype(int)
    return {
        "Precision (%)": precision_score(actual, predicted, zero_division=0) * 100,
        "Recall (%)": recall_score(actual, predicted, zero_division=0) * 100,
        "F1-Score (%)": f1_score(actual, predicted, zero_division=0) * 100,
        "False Alarms": int(((actual == 0) & (predicted == 1)).sum()),
        "Missed Failures": int(((actual == 1) & (predicted == 0)).sum()),
        "Machines Flagged": int(predicted.sum())
    }

actual = validation_results["Actual Failure"].astype(int).to_numpy()
original_alerts = (
    validation_results["Agent Decision"] != "NORMAL - Continue Operation"
).astype(int).to_numpy()

threshold_results = []
base = evaluate_alerts(actual, original_alerts)
base["Rule"] = "Original Agent"
threshold_results.append(base)

for threshold in [0.10, 0.20, 0.30, 0.40, 0.50, 0.60]:
    candidate_alerts = original_alerts.copy()
    escalate = (
        (validation_results["Torque_Trend_State"].to_numpy() == "Increasing") &
        (validation_results["Failure Probability"].to_numpy() >= threshold)
    )
    candidate_alerts[escalate] = 1
    metrics = evaluate_alerts(actual, candidate_alerts)
    metrics["Rule"] = f"Increasing trend + probability >= {threshold:.0%}"
    threshold_results.append(metrics)

threshold_comparison = pd.DataFrame(threshold_results).set_index("Rule")
print("\n===== VALIDATION: TREND ESCALATION THRESHOLD COMPARISON =====")
display(threshold_comparison.round(2))

# ================================================================
# 11. POLICY EVALUATION AND FEEDBACK-BASED SELECTION
# ================================================================

MISSED_FAILURE_COST = 10.0
FALSE_ALARM_COST = 1.0
MONITORING_COST = 0.1
MAINTENANCE_COST = 0.5
CRITICAL_COST = 1.0


def evaluate_policy(data, policy_name):
    probabilities = data["Failure Probability"].to_numpy()
    actual_values = data["Actual Failure"].to_numpy().astype(int)
    actions = make_decisions(probabilities, POLICIES[policy_name])

    alerts = (actions != "Normal").astype(int)
    tn, fp, fn, tp = confusion_matrix(
        actual_values, alerts, labels=[0, 1]
    ).ravel()

    monitoring = int((actions == "Monitor").sum())
    maintenance = int((actions == "Schedule Maintenance").sum())
    critical = int((actions == "Critical").sum())

    total_cost = (
        fn * MISSED_FAILURE_COST +
        fp * FALSE_ALARM_COST +
        monitoring * MONITORING_COST +
        maintenance * MAINTENANCE_COST +
        critical * CRITICAL_COST
    )

    return {
        "Policy": policy_name,
        "Precision (%)": precision_score(actual_values, alerts, zero_division=0) * 100,
        "Recall (%)": recall_score(actual_values, alerts, zero_division=0) * 100,
        "F1 Score (%)": f1_score(actual_values, alerts, zero_division=0) * 100,
        "False Alarms": int(fp),
        "Missed Failures": int(fn),
        "Normal": int((actions == "Normal").sum()),
        "Monitor": monitoring,
        "Schedule Maintenance": maintenance,
        "Critical": critical,
        "Total Cost": float(total_cost),
        "Cost per Machine": float(total_cost / len(data))
    }

policy_rows = [evaluate_policy(validation_results, p) for p in POLICIES]
policy_comparison = pd.DataFrame(policy_rows).sort_values("Total Cost")

print("\n===== VALIDATION POLICY COMPARISON =====")
display(policy_comparison.round(3))

best_f1_policy = policy_comparison.sort_values("F1 Score (%)", ascending=False).iloc[0]["Policy"]
best_cost_policy = policy_comparison.sort_values("Total Cost", ascending=True).iloc[0]["Policy"]

print("Best F1 policy:", best_f1_policy)
print("Best cost policy:", best_cost_policy)

# ================================================================
# 12. SEQUENTIAL SELF-EVOLVING AGENT SIMULATION
# ================================================================

candidate_policies = list(POLICIES.keys())
initial_policy = "Original (30%, 60%, 80%)"
BATCH_SIZE = 400

feedback_data = validation_results.copy().reset_index(drop=True)
batches = [
    feedback_data.iloc[i:i + BATCH_SIZE].copy()
    for i in range(0, len(feedback_data), BATCH_SIZE)
]

current_policy = initial_policy
observed_feedback = []
adaptive_history = []

for batch_number, batch in enumerate(batches, start=1):
    batch_result = evaluate_policy(batch, current_policy)
    adaptive_history.append({
        "Batch": batch_number,
        "Policy Applied": current_policy,
        "Machines": len(batch),
        "Failures": int(batch["Actual Failure"].sum()),
        "Precision (%)": batch_result["Precision (%)"],
        "Recall (%)": batch_result["Recall (%)"],
        "F1 Score (%)": batch_result["F1 Score (%)"],
        "False Alarms": batch_result["False Alarms"],
        "Missed Failures": batch_result["Missed Failures"],
        "Total Cost": batch_result["Total Cost"],
        "Cost per Machine": batch_result["Cost per Machine"]
    })

    observed_feedback.append(batch.copy())
    past_data = pd.concat(observed_feedback, ignore_index=True)

    candidate_results = pd.DataFrame([
        evaluate_policy(past_data, p) for p in candidate_policies
    ])
    current_policy = candidate_results.loc[
        candidate_results["Total Cost"].idxmin(), "Policy"
    ]

adaptive_history_df = pd.DataFrame(adaptive_history)
print("\n===== SELF-EVOLVING AGENT =====")
display(adaptive_history_df.round(3))
print("Policy selected after all feedback:", current_policy)

# Fair comparison over the same validation records.
comparison_rows = []

total_failures = int(feedback_data["Actual Failure"].sum())
total_machines = len(feedback_data)
adaptive_fp = int(adaptive_history_df["False Alarms"].sum())
adaptive_fn = int(adaptive_history_df["Missed Failures"].sum())
adaptive_cost = float(adaptive_history_df["Total Cost"].sum())
adaptive_tp = total_failures - adaptive_fn
adaptive_precision = adaptive_tp / (adaptive_tp + adaptive_fp) if adaptive_tp + adaptive_fp else 0
adaptive_recall = adaptive_tp / total_failures if total_failures else 0
adaptive_f1 = 2 * adaptive_precision * adaptive_recall / (adaptive_precision + adaptive_recall) if adaptive_precision + adaptive_recall else 0

comparison_rows.append({
    "Evaluation": "Self-Evolving Agent",
    "Precision (%)": adaptive_precision * 100,
    "Recall (%)": adaptive_recall * 100,
    "F1 Score (%)": adaptive_f1 * 100,
    "False Alarms": adaptive_fp,
    "Missed Failures": adaptive_fn,
    "Total Cost": adaptive_cost,
    "Cost per Machine": adaptive_cost / total_machines
})

for policy_name in candidate_policies:
    r = evaluate_policy(feedback_data, policy_name)
    comparison_rows.append({
        "Evaluation": "Fixed: " + policy_name,
        "Precision (%)": r["Precision (%)"],
        "Recall (%)": r["Recall (%)"],
        "F1 Score (%)": r["F1 Score (%)"],
        "False Alarms": r["False Alarms"],
        "Missed Failures": r["Missed Failures"],
        "Total Cost": r["Total Cost"],
        "Cost per Machine": r["Cost per Machine"]
    })

evolution_comparison_df = pd.DataFrame(comparison_rows).sort_values("Total Cost")
print("\n===== SELF-EVOLVING AGENT VS FIXED POLICIES =====")
display(evolution_comparison_df.round(3))

# ================================================================
# 13. REPEATED BATCH-ORDER ROBUSTNESS
# ================================================================

N_REPEATS = 10
experiment_rows = []

for seed in range(N_REPEATS):
    shuffled_data = feedback_data.sample(frac=1, random_state=seed).reset_index(drop=True)
    shuffled_batches = [
        shuffled_data.iloc[i:i + BATCH_SIZE].copy()
        for i in range(0, len(shuffled_data), BATCH_SIZE)
    ]

    current_policy = initial_policy
    observed = []
    adaptive_fp = adaptive_fn = 0
    adaptive_cost = 0.0

    for batch in shuffled_batches:
        r = evaluate_policy(batch, current_policy)
        adaptive_fp += r["False Alarms"]
        adaptive_fn += r["Missed Failures"]
        adaptive_cost += r["Total Cost"]
        observed.append(batch.copy())
        past = pd.concat(observed, ignore_index=True)
        candidates = pd.DataFrame([evaluate_policy(past, p) for p in candidate_policies])
        current_policy = candidates.loc[candidates["Total Cost"].idxmin(), "Policy"]

    failures = int(shuffled_data["Actual Failure"].sum())
    tp = failures - adaptive_fn
    p = tp / (tp + adaptive_fp) if tp + adaptive_fp else 0
    r = tp / failures if failures else 0
    f1 = 2 * p * r / (p + r) if p + r else 0

    experiment_rows.append({
        "Repeat": seed + 1,
        "Policy": "Self-Evolving Agent",
        "Precision (%)": p * 100,
        "Recall (%)": r * 100,
        "F1 Score (%)": f1 * 100,
        "False Alarms": adaptive_fp,
        "Missed Failures": adaptive_fn,
        "Total Cost": adaptive_cost,
        "Cost per Machine": adaptive_cost / len(shuffled_data)
    })

    for policy_name in candidate_policies:
        r_fixed = evaluate_policy(shuffled_data, policy_name)
        experiment_rows.append({
            "Repeat": seed + 1,
            "Policy": "Fixed: " + policy_name,
            "Precision (%)": r_fixed["Precision (%)"],
            "Recall (%)": r_fixed["Recall (%)"],
            "F1 Score (%)": r_fixed["F1 Score (%)"],
            "False Alarms": r_fixed["False Alarms"],
            "Missed Failures": r_fixed["Missed Failures"],
            "Total Cost": r_fixed["Total Cost"],
            "Cost per Machine": r_fixed["Cost per Machine"]
        })

repeated_results_df = pd.DataFrame(experiment_rows)
reliability_summary = repeated_results_df.groupby("Policy")[
    ["Precision (%)", "Recall (%)", "F1 Score (%)", "False Alarms", "Missed Failures", "Total Cost", "Cost per Machine"]
].agg(["mean", "std"])

print("\n===== REPEATED BATCH-ORDER EXPERIMENT =====")
display(reliability_summary.round(3))

# ================================================================
# 14. FEATURE ABLATION: 5-FOLD CV
# ================================================================

ablation_sets = {
    "All Features": list(X_train.columns),
    "Torque + Tool Wear": [c for c in X_train.columns if c in ["Torque__Nm_", "Tool_wear__min_"]]
}

# If cleaned names differ, identify them robustly.
torque_feature = next((c for c in X_train.columns if "Torque" in c), None)
tool_feature = next((c for c in X_train.columns if "Tool_wear" in c), None)
if torque_feature and tool_feature:
    ablation_sets["Torque + Tool Wear"] = [torque_feature, tool_feature]

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
ablation_rows = []

for set_name, columns in ablation_sets.items():
    fold_metrics = []
    for train_idx, val_idx in cv.split(X_train[columns], y_train):
        Xf = X_train.iloc[train_idx][columns]
        Xv = X_train.iloc[val_idx][columns]
        yf = y_train.iloc[train_idx]
        yv = y_train.iloc[val_idx]
        pos = int((yf == 1).sum())
        neg = int((yf == 0).sum())

        m = XGBClassifier(
            n_estimators=300,
            max_depth=5,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="binary:logistic",
            eval_metric="logloss",
            scale_pos_weight=neg / max(pos, 1),
            random_state=42,
            n_jobs=-1
        )
        m.fit(Xf, yf)
        pr = m.predict_proba(Xv)[:, 1]
        pdx = (pr >= 0.5).astype(int)
        fold_metrics.append({
            "Precision": precision_score(yv, pdx, zero_division=0),
            "Recall": recall_score(yv, pdx, zero_division=0),
            "F1": f1_score(yv, pdx, zero_division=0),
            "ROC-AUC": roc_auc_score(yv, pr)
        })

    fold_df = pd.DataFrame(fold_metrics)
    ablation_rows.append({
        "Feature Set": set_name,
        "Features": len(columns),
        "Precision Mean": fold_df["Precision"].mean(),
        "Recall Mean": fold_df["Recall"].mean(),
        "F1 Mean": fold_df["F1"].mean(),
        "ROC-AUC Mean": fold_df["ROC-AUC"].mean(),
        "F1 Std": fold_df["F1"].std()
    })

ablation_results = pd.DataFrame(ablation_rows)
print("\n===== FEATURE ABLATION =====")
display(ablation_results.round(4))

# ================================================================
# 15. REPEATED STRATIFIED CROSS-VALIDATION ROBUSTNESS
# ================================================================

cv_policies = {
    "Original": POLICIES["Original (30%, 60%, 80%)"],
    "Best F1": POLICIES["Best F1 (35%, 65%, 85%)"],
    "Best Cost": POLICIES["Best Cost (25%, 55%, 80%)"]
}

robustness_rows = []
X_cv = X_train.copy()
y_cv = np.asarray(y_train)

for repeat, seed in enumerate([11, 21, 31], start=1):
    splitter = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
    oof_prob = np.zeros(len(y_cv))

    for train_idx, val_idx in splitter.split(X_cv, y_cv):
        Xf, Xv = X_cv.iloc[train_idx], X_cv.iloc[val_idx]
        yf = y_cv[train_idx]
        pos = int((yf == 1).sum())
        neg = int((yf == 0).sum())

        m = XGBClassifier(
            n_estimators=300,
            max_depth=5,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="binary:logistic",
            eval_metric="logloss",
            scale_pos_weight=neg / max(pos, 1),
            random_state=seed,
            n_jobs=-1
        )
        m.fit(Xf, yf)
        oof_prob[val_idx] = m.predict_proba(Xv)[:, 1]

    for policy_name, thresholds in cv_policies.items():
        actions = make_decisions(oof_prob, thresholds)
        alerts = (actions != "Normal").astype(int)
        robustness_rows.append({
            "Repeat": repeat,
            "Policy": policy_name,
            "ROC-AUC": roc_auc_score(y_cv, oof_prob),
            "Precision": precision_score(y_cv, alerts, zero_division=0),
            "Recall": recall_score(y_cv, alerts, zero_division=0),
            "F1": f1_score(y_cv, alerts, zero_division=0),
            "False Alarms": int(((y_cv == 0) & (alerts == 1)).sum()),
            "Missed Failures": int(((y_cv == 1) & (alerts == 0)).sum())
        })

robustness_results = pd.DataFrame(robustness_rows)
robustness_summary = robustness_results.groupby("Policy").agg(
    Precision_Mean=("Precision", "mean"),
    Precision_Std=("Precision", "std"),
    Recall_Mean=("Recall", "mean"),
    Recall_Std=("Recall", "std"),
    F1_Mean=("F1", "mean"),
    F1_Std=("F1", "std"),
    ROC_AUC_Mean=("ROC-AUC", "mean"),
    ROC_AUC_Std=("ROC-AUC", "std")
).reset_index()

print("\n===== REPEATED STRATIFIED CV ROBUSTNESS =====")
display(robustness_summary.round(4))

plt.figure(figsize=(8, 5))
plt.bar(
    robustness_summary["Policy"],
    robustness_summary["F1_Mean"],
    yerr=robustness_summary["F1_Std"],
    capsize=5
)
plt.ylabel("Mean F1-score")
plt.title("Repeated Stratified CV: F1-score")
plt.ylim(0, 1)
plt.tight_layout()
plt.savefig("repeated_cv_f1_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

# ================================================================
# 16. FINAL PROJECT SUMMARY
# ================================================================

main_metrics = {
    "Accuracy (%)": accuracy_score(y_test, y_pred) * 100,
    "Precision (%)": precision_score(y_test, y_pred, zero_division=0) * 100,
    "Recall (%)": recall_score(y_test, y_pred, zero_division=0) * 100,
    "F1 (%)": f1_score(y_test, y_pred, zero_division=0) * 100,
    "ROC-AUC (%)": roc_auc_score(y_test, failure_probability) * 100
}

print("\n" + "=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)
print("Dataset:", df.shape[0], "records")
print("Predictor count:", X_train.shape[1])
print("Main XGBoost metrics:")
for k, v in main_metrics.items():
    print(f"  {k}: {v:.2f}")
print("Best F1 validation policy:", best_f1_policy)
print("Best cost validation policy:", best_cost_policy)
print("Self-evolving final selected policy:", current_policy)
print("\nPROJECT PIPELINE COMPLETED SUCCESSFULLY")
```
